```python
# Renaming the column

df.rename(columns={'Selling_Price_Lakh_INR' : 'Price'}, inplace = True)
df.head()
```

```python
#   Encoding of Catogrical columns 

df = pd.get_dummies(df,columns=['Brand','Model','Body_Type','Fuel_Type',
        'Transmission','Owner_Type'],drop_first=True, dtype=int)
df
```

```python
#  Droping of unwanted Column 

df = df.drop('Car_ID', axis=1)
df
```

```python
# Importing input and output columns in variable.

X = df.drop('Price', axis=1)
y = df['Price']
```

```python
#  Training & Teasting of columns.

from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(X,y, test_size=0.2, random_state=2)
```

```python
#  applying polynomial feature in columns which do not have consistancy in data.

from sklearn.preprocessing import PolynomialFeatures, StandardScaler

poly_cols = ["Kms_Driven", "Engine_CC", "Power_BHP", "Car_Age_Years"]

# creating object of the model
poly = PolynomialFeatures(degree=2)


#  fitting the data in model

poly_train = pd.DataFrame(poly.fit_transform(X_train[poly_cols]),columns= poly.get_feature_names_out(poly_cols), index=X_train.index)
poly_test = pd.DataFrame(poly.transform(X_test[poly_cols]), columns=poly.get_feature_names_out(poly_cols), index= X_test.index)


# droping of old data and concatination of new data gotted by fitting

X_train2= pd.concat([X_train.drop(columns=poly_cols), poly_train], axis=1)
X_test2 = pd.concat([X_test.drop(columns=poly_cols),poly_test],axis=1)


# scaling of data to make the low difference between data.

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train2)
X_test_scaled = scaler.fit_transform(X_test2)


# importing linear regression for prediction
from sklearn.linear_model import LinearRegression

# creating object of the model
model = LinearRegression()

#  fitting the data in model
model.fit(X_train_scaled,y_train)

# prediction of the model
predict = model.predict(X_test_scaled)


# checking the accuracy of working model 
from sklearn.metrics import r2_score

r2 = r2_score(predict,y_test)
r2
```

```python
# Creating a streamlit file

import joblib
joblib.dump(model,'car_price_model.pkl')
joblib.dump(scaler, 'scaler.pkl')
joblib.dump(X.columns.to_list(), 'columns.pkl')
```
